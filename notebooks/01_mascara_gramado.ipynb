# 01 — Máscara do gramado
Limiarização em HSV (sistemas de cores, operação pontual) + morfologia. Limiares em `configs/default.yaml`, justificados pelo histograma do canal H.

```python
import os, sys
sys.path.insert(0, '..')
import cv2, yaml
import matplotlib.pyplot as plt
from src.io_video import sample_frames
from src.field_mask import hue_histogram, debug_view

CLIP = '../data/clips/clip01.mp4'
OUT = '../outputs/figures'
os.makedirs(OUT, exist_ok=True)
cfg = yaml.safe_load(open('../configs/default.yaml'))['field_mask']
frames = sample_frames(CLIP, 10)
```

```python
hist = hue_histogram(frames, cfg['s_min'], cfg['v_min'])
fig, ax = plt.subplots(figsize=(8, 3))
ax.bar(range(180), hist / hist.sum(), width=1, color=[plt.cm.hsv(h / 180) for h in range(180)])
for h in cfg['h_range']:
    ax.axvline(h, color='k', ls='--')
ax.set_xlabel('H (OpenCV, 0–179)'); ax.set_ylabel('fração dos pixels')
ax.set_title(f"Histograma do canal H (S≥{cfg['s_min']}, V≥{cfg['v_min']}) — faixa escolhida {cfg['h_range']}")
fig.tight_layout(); fig.savefig(f'{OUT}/01_histograma_h.png', dpi=150)
```

```python
panel = debug_view(frames[0], cfg)
cv2.imwrite(f'{OUT}/01_mascara_gramado.png', panel)
plt.figure(figsize=(12, 7)); plt.imshow(cv2.cvtColor(panel, cv2.COLOR_BGR2RGB)); plt.axis('off')
```
